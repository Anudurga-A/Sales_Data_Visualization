```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Month': ['January', 'February', 'March', 'April',
              'May', 'June'],
    'Sales': [12000, 15000, 10000, 18000, 22000, 25000]
}

df = pd.DataFrame(data)

print(df)
```

```python
plt.figure(figsize=(10, 5))

plt.bar(df['Month'], df['Sales'])

plt.title('Monthly Sales Comparison')
plt.xlabel('Month')
plt.ylabel('Sales Amount')

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(df['Month'], df['Sales'], marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales Amount')

plt.grid(True)
plt.show()
```

Observations
Sales increased from ₹12,000 in January to ₹25,000 in June, showing an overall upward trend.
The highest sales were recorded in June, while the lowest sales were recorded in March.
